Numpy Array Operations:
Scenario: You are given a list of daily average temperatures (in degrees Celsius) recorded over two
weeks. Your task is to use NumPy to analyze this data.
1. Create a 1D NumPy Array:

Create a 1D NumPy array named temperatures_w1 for Week 1 with the following
values:
[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]

2. Inspection and Properties:

● Inspect the shape, data type, and the number of elements in the array. Inspect
the shape of the temperatures_w1 array.

3. Array Operations:

● Convert the temperatures_w1 from Celsius to Fahrenheit using the formula:
Fahrenheit = (Celsius * 9/5) + 32
● Find the maximum, minimum, and mean temperatures for the week.

4. Array Slicing and Indexing:

● Extract the temperatures for the first three days of the week.
● Extract the temperatures for the weekend (last two days).
● Extract the temperatures for the middle three days of the week.

5. Create a 2D Array:

Create a 2D NumPy array named temperatures where each row represents
temperatures for a different week:
★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)
★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

6. Inspect and Slice the 2D Array:

● Inspect and print the shape, data type, and total number of elements in the
temperatures array.
● Extract the temperatures for each week and the weekends (last two days) for
both weeks.

```python
#1. Create a 1D NumPy Array
#First, import NumPy and create an array named temperatures_w1.

import numpy as np

# Week 1 temperatures in Celsius
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:", temperatures_w1)
```

```python
#2. Inspection and Properties
#shape – Returns the dimensions of the array.

#dtype – Returns the data type of elements.

#size – Returns the total number of elements.

print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
#3. Array Operations
#A. Convert Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", temperatures_f)
```

```python
#B. Find maximum, minimum, and mean temperatures

print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
#4. Array Slicing and Indexing
#A. First three days
#Python uses zero-based indexing, and the ending index in slicing is excluded

print("First Three Days:", temperatures_w1[:3])
```

```python
#B. Weekend (last two days)

print("Weekend Temperatures:", temperatures_w1[-2:])
```

```python
#C. Middle three days

print("Middle Three Days:", temperatures_w1[2:5])
```

```python
#5. Create a 2D NumPy Array
#Create a 2D array named temperatures, where each row represents one week's temperatures.

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Two Weeks Temperatures:")
print(temperatures)
```

```python
#6. Inspect and Slice the 2D Array
#A. Inspect array properties

print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Number of Elements:", temperatures.size)
```

```python
#B. Extract temperatures for each week

# Week 1 temperatures
print("Week 1:", temperatures[0])

# Week 2 temperatures
print("Week 2:", temperatures[1])
```

```python
#C. Extract weekend temperatures for both weeks
# Use [:, -2:] to extract the last two columns from all rows.

weekend_temperatures = temperatures[:, -2:]

print("Weekend Temperatures for Both Weeks:")
print(weekend_temperatures)
```

Pandas Series:

1. Creating Pandas Series:
● Create a Pandas Series named marks with the following data:
➢ Marks: 95, 92, 89, 85, 80
➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

2. Indexing and Slicing:
● Use the integer index position to access the mark of the 1st rank student.
● Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index
labels.
● Use the iloc accessor to retrieve the mark of the 3rd-ranked student.
● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are
greater than 90.
3. Manipulating Series:
● Modify the mark of the student with the 1st rank to 100.
● Remove the entry corresponding to the last rank from the Series.
● Compute the CGPA by dividing each mark in the Series by 10.

```python
#Pandas Series
#1. Creating a Pandas Series
#First, import Pandas and create a Series named marks with the given marks and custom indices.

import pandas as pd

# Create a Pandas Series
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

```python
#2. Indexing and Slicing
#A. Access the 1st rank student's mark using integer position
#Use integer indexing with iloc[0].

print("1st Rank Mark:", marks.iloc[0])
```

```python
#B. Retrieve the top 3 ranks using loc
#The loc accessor retrieves values using index labels.

print("Top 3 Rank Marks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

```python
#C. Retrieve the 3rd-ranked student's mark using iloc
#The iloc accessor retrieves values using integer positions (starting from 0).

print("3rd Rank Mark:", marks.iloc[2])
```

```python
#D. Apply Boolean Masking
#Retrieve the ranks where marks are greater than 90.

print("Marks Greater Than 90:")
print(marks[marks > 90])
```

```python
#3. Manipulating Pandas Series
#A. Modify the 1st rank student's mark to 100
#Use loc to modify the value.

marks.loc['Rank1'] = 100

print("Updated Marks:")
print(marks)
```

```python
#B. Remove the last rank from the Series
#Use the drop() method to remove Rank5.

marks = marks.drop('Rank5')

print("Marks After Removing Last Rank:")
print(marks)
```

```python
#C. Compute CGPA by dividing each mark by 10
#Perform vectorized division on the Series.

cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

Pandas DataFrame:

1. Creating Pandas DataFrame:

● Create a Pandas DataFrame named transactions with the following data: ➢
TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',

'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South',
'North'
➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

2. Data Exploration:
● Display the transactions DataFrame and its basic information, including head, tail,
shape, column names, and data types.
● Display only the 'ProductCategory' and 'Amount' columns.
● Retrieve the last 3 columns of the DataFrame using iloc or loc.
● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
● Find the value counts for the 'ProductCategory' column.
● Find the unique values in the 'Region' column.
● Group by 'Region' and find the mean amount for each region.

3. Manipulating the DataFrame:
● Modify the 'Amount' for the transaction with TransactionID 102 to 165.
● Add a new column 'Discount' by calculating 10% of the 'Amount'.
● Remove the row with TransactionID 109.
● Delete the 'Discount' column from the DataFrame.

```python
#Pandas DataFrame – Transaction Analysis
#1. Creating a Pandas DataFrame
#First, import Pandas and create a DataFrame named transactions using the given transaction details.

import pandas as pd

# Create DataFrame
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print("Transactions DataFrame:")
print(transactions)
```

```python
#2. Data Exploration
#A. Display the DataFrame and basic information
#1. Display the complete DataFrame

print(transactions)
```

```python
#2. Display the first five rows using head()

print("First 5 Rows:")
print(transactions.head())
```

```python
#3. Display the last five rows using tail()

print("Last 5 Rows:")
print(transactions.tail())
```

```python
#4. Display the shape

print("Shape:", transactions.shape)
```

```python
# 5. Display column names

print("Column Names:", transactions.columns)
```

```python
#6. Display data types

print("Data Types:")
print(transactions.dtypes)
```

```python
#7. Display basic information

transactions.info()
```

```python
#B. Display only ProductCategory and Amount columns
#Use double square brackets to select multiple columns.

print(transactions[['ProductCategory', 'Amount']])
```

```python
#C. Retrieve the last three columns using iloc
#Use iloc[:, -3:] to retrieve all rows and the last three columns.

print(transactions.iloc[:, -3:])
```

```python
#D. Filter rows where Region is North and Amount is greater than 200
#Use Boolean conditions with the & operator.

filtered_data = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(filtered_data)
```

```python
#E. Find value counts for ProductCategory
#Use value_counts() to count the occurrences of each product category.

print(transactions['ProductCategory'].value_counts())
```

```python
#F. Find unique values in Region
#Use unique() to display the distinct regions.

print("Unique Regions:")
print(transactions['Region'].unique())
```

```python
#G. Group by Region and calculate mean Amount
#Use groupby() and mean() to calculate the average transaction amount for each region.

print("Average Amount by Region:")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
#3. Manipulating the DataFrame
#A. Modify Amount for TransactionID 102 to 165
#Use loc to find the transaction and update its amount.

transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

print(transactions)
```

```python
#B. Add a new Discount column (10% of Amount)
#Multiply the Amount column by 0.10.

transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

```python
#C. Remove the row with TransactionID 109
#Use drop() to remove the row based on its index.

transactions = transactions.drop(
    transactions[transactions['TransactionID'] == 109].index
)

print(transactions)
```

```python
#D. Delete the Discount column
#Use drop() with axis=1 to remove a column.

transactions = transactions.drop('Discount', axis=1)

print("Final DataFrame:")
print(transactions)
```
